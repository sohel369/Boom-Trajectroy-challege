# Asteroid Data Analysis

In this notebook, we evaluate the quality of our data, check for missing values, and visualize the relationships between features.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load data
df = pd.read_csv('../data/raw/asteroid_data.csv')
print("Data Loaded!")
```

## 1. Check Missing Values

```python
# Summing all null values per column
missing_data = df.isnull().sum()
print(missing_data[missing_data > 0])
```

## 2. Basic Statistics

```python
# Summary statistics for numerical columns
df.describe()
```

## 3. Visualizing Data

```python
# Plot 1: Target Distribution (Hazardous vs Non-Hazardous)
plt.figure(figsize=(6, 4))
sns.countplot(x='Hazardous', data=df, palette='viridis')
plt.title('Count of Hazardous vs Non-Hazardous Asteroids')
plt.show()
```

```python
# Plot 2: Absolute Magnitude (Brightness) vs Hazardous
plt.figure(figsize=(10, 6))
sns.boxplot(x='Hazardous', y='Absolute Magnitude', data=df)
plt.title('Brightness (Magnitude) vs Hazard Status')
plt.show()
```

## 4. Is Cleaning Needed?

Based on the analysis:

1. **Missing Values**: None! The dataset is very clean in this regard.
2. **Redundancy**: There are multiple columns for the same information in different units (KM, Meters, Miles, Feet). We should **drop** all but one unit (e.g., KM).
3. **Identifiers**: 'Neo Reference ID' and 'Name' don't help the model learn physics. They should be removed before training.
4. **Class Imbalance**: Only about ~16% of the asteroids are 'Hazardous'. Your model might need special care (like 'class weighting') to learn them effectively.
