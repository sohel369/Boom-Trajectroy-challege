In this notebook, we will load the NASA asteroid dataset, explore its features, and prepare our data for machine learning.

## 4. Column Explanations

- **Neo Reference ID**: Unique ID for the object.
- **Absolute Magnitude**: How bright the asteroid is (smaller = bigger).
- **Est Dia (KM)**: Size range of the asteroid.
- **Relative Velocity**: Speed relative to Earth.
- **Miss Dist (kilometers)**: Distance it bypasses Earth.
- **Orbit Uncertainity**: Scale 0-9 (high = very unsure path).
- **Hazardous**: Our TARGET variable (True/False).

## 5. Separating X and y

```python
# Select numerical features that influence impact hazard
features = [
    'Absolute Magnitude', 
    'Est Dia in KM(min)', 
    'Est Dia in KM(max)', 
    'Relative Velocity km per sec', 
    'Miss Dist.(kilometers)', 
    'Orbit Uncertainity', 
    'Minimum Orbit Intersection'
]

X = df[features]
y = df['Hazardous']

print(f"X Shape: {X.shape}")
print(f"y Shape: {y.shape}")
```
